```python
import tensorflow as tf
```

```python
# 1. Load your existing Keras model (or build a new one)
# Keras 3 typically uses the .keras extension, but .h5 works as well.
model = tf.keras.models.load_model('./facenet.keras')
```

```python
# 2. Initialize the LiteRT Converter
converter = tf.lite.TFLiteConverter.from_keras_model(model)

# 3. (Optional but Recommended) Apply optimizations
# This enables post-training quantization to reduce model size and improve latency
# with minimal impact on accuracy.
converter.optimizations = [tf.lite.Optimize.DEFAULT]

# 4. Convert the model
litert_model = converter.convert()
```

```python
# 5. Save the converted model to disk
with open('converted_model.tflite', 'wb') as f:
    f.write(litert_model)
```
